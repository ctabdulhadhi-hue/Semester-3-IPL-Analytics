```python
import sqlite3
import pandas as pd

con = sqlite3.connect('C:\\Users\\ctabd\\OneDrive\\Desktop\\3rd Sem\\project_3.1\\2_IPL_Cricket\\ipl.db')

def q(sql):
    return pd.read_sql_query(sql, con)
```

```python
def run_sql_file(path):
    with open(path,'r')as f:
        script=f.read()
    con.executescript(script)
    con.commit()
    print('ran',path)
```

```python
q("""
-- What is the strike rate of every batter in the death overs

SELECT
    batter,
    ROUND(
        100.0 * SUM(batter_runs) /
        SUM(
            CASE
                WHEN is_wide_ball = 0
                 AND is_no_ball = 0
                THEN 1
                ELSE 0
            END
        ),
        2
    ) AS strike_rate
FROM deliveries
WHERE over_number >= 15
  AND is_super_over = 0
GROUP BY batter;
""")

```
